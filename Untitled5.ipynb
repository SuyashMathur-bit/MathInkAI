```python
def extract_ground_truth(file_path):

    tree = ET.parse(file_path)
    root = tree.getroot()

    for element in root.iter():

        if element.tag.endswith("annotation"):

            if element.attrib.get("type") == "truth":

                if element.text is not None:

                    value = element.text.strip()

                    if value != "Segmentation":

                        return value

    return None
```

```python
def extract_strokes(file_path):

    tree = ET.parse(file_path)

    root = tree.getroot()

    strokes = []

    for trace in root.iter():

        if trace.tag.endswith("trace"):

            if trace.text is None:
                continue

            points = []

            for point in trace.text.strip().split(","):

                values = point.strip().split()

                if len(values) >= 2:

                    x = float(values[0])
                    y = float(values[1])

                    points.append((x, y))

            if points:
                strokes.append(points)

    return strokes
```

```python
def normalize_strokes(strokes):

    x_values = []
    y_values = []

    for stroke in strokes:
        for x, y in stroke:
            x_values.append(x)
            y_values.append(y)

    min_x = min(x_values)
    max_x = max(x_values)

    min_y = min(y_values)
    max_y = max(y_values)

    width = max_x - min_x
    height = max_y - min_y

    if width == 0:
        width = 1

    if height == 0:
        height = 1

    normalized_strokes = []

    for stroke in strokes:

        normalized_stroke = []

        for x, y in stroke:

            new_x = (x - min_x) / width
            new_y = (y - min_y) / height

            normalized_stroke.append((new_x, new_y))

        normalized_strokes.append(normalized_stroke)

    return normalized_strokes
       
```

```python
def process_dataset(files):

    processed_data = []

    for file in files:

        strokes = extract_strokes(file)

        if len(strokes) == 0:
            continue

        label = extract_ground_truth(file)

        if label is None:
            continue

        normalized_strokes = normalize_strokes(strokes)
        
        sample = {
            "file": file,
            "strokes": normalized_strokes,
            "label": label
        }

        processed_data.append(sample)

    return processed_data
```

```python
import re
def tokenize_latex(expression):
    expression=expression.replace("$","")
    tokens=re.findall(
         r'\\[a-zA-Z]+|[0-9]+|[a-zA-Z]|[^\s]',
        expression
    )

    return tokens
```

```python
test_labels = [
    r"$\phi(x)$",
    r"$x^2$",
    r"$\frac{x}{2}$"
]

for label in test_labels:
    print(label, "→", tokenize_latex(label))
```

```python
from collections import Counter
token_counter=Counter()
for sample in data:
    tokens=tokenize_latex(sample["label"])
    token_counter.update(tokens)
    print("Total unique tokens:", len(token_counter))

print("\nMost common tokens:")
for token, count in token_counter.most_common(30):
    print(token, ":", count)
```

```python
special_tokens=[
    "<blank>",
    "<pad>",
    "<unk>"
]
unique_tokens=sorted(token_counter.keys())
vocabulary=special_tokens+unique_tokens
print("Vocabulary size:", len(vocabulary))
print("\nFirst 30 tokens:")

for i, token in enumerate(vocabulary[:30]):
    print(i, ":", token)
```

```python
token_to_id={
    token:i
    for i, token in enumerate(vocabulary)
}

id_to_token={
    i:token
    for i,token in enumerate(vocabulary)
}
print("Token to ID:")
print(token_to_id)

print("\nID to Token:")
print(id_to_token)
```

```python
def encode_label(label):
    token=tokenize_latex(label)
    encoded=[]
    for token in tokens:
        if token in token_to_id:
            encoded.append(token_to_id[token])
        else:
            encoded.append(token_to_id["<unk>"])

    return encoded
```

```python
import numpy as np
def strokes_to_sequence(strokes):
    sequence=[]
    for stoke_index,stroke in enumerate(strokes):
        for point_index,(x,y) in enumerate(stroke):
            pen=1 if point_index==0 else 0
            sequence.append([x,y,pen])

        return np.array(sequence, dtype=np.float32)
```

```python
import tensorflow as tf 
from tensorflow.keras import layers,Model

stroke_input=layers.Input(
    shape=(max_length,3),
    name="stroke_input"
)
x=layers.Bidirectional(
    layers.LSTM(
        128,
        return_sequences="True"
    )
)(stroke_input)

x = layers.Bidirectional(
    layers.LSTM(
        128,
        return_sequences=True
    )
)(x)

token_output=layers.Dense(
    len(vocabulary),
    activation="softmax",
    name="token_output"
)(x)


labels = layers.Input(
    shape=(max_label_length,),
    dtype="int32",
    name="labels"
)
input_length=layers.Input(
    shape=(1,),
    dtype="int32",
    name="input_length"
)
label_length=layers.Input(
    shape=(1,),
    dtype="int32",
    name="label_length"
)

```

```python
def ctc_loss_function(args):
    labels, y_pred, input_length, label_length = args

    return tf.keras.backend.ctc_batch_cost(
        labels,
        y_pred,
        input_length,
        label_length
    )

loss_output=layers.Lambda(
    ctc_loss_function,
    name="ctc_loss",
)([
    labels,
    token_output,
    input_length,
    label_length
])
ctc_model = Model(
    inputs=[
        stroke_input,
        labels,
        input_length,
        label_length
    ],
    outputs=loss_output
)
ctc_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss=lambda y_true, y_pred: y_pred
)
ctc_model.summary()
```

```python
import numpy as np
input_lengths_train_ctc = input_lengths_train.reshape(-1, 1)
input_lengths_val_ctc = input_lengths_val.reshape(-1, 1)

label_lengths_train_ctc = label_lengths_train.reshape(-1, 1)
label_lengths_val_ctc = label_lengths_val.reshape(-1, 1)

y_train_ctc=y_train_padded.copy()
y_val_ctc=y_val_padded.copy()

y_train_ctc[y_train_ctc == -1] = 0
y_val_ctc[y_val_ctc == -1] = 0

train_inputs={
     "stroke_input": X_val_padded,
    "labels": y_val_ctc,
    "input_length": input_lengths_val_ctc,
    "label_length": label_lengths_val_ctc
}

val_inputs = {
    "stroke_input": X_val_padded,
    "labels": y_val_ctc,
    "input_length": input_lengths_val_ctc,
    "label_length": label_lengths_val_ctc
}
train_dummy = np.zeros(
    (len(X_train), 1),
    dtype=np.float32
)

val_dummy = np.zeros(
    (len(X_val), 1),
    dtype=np.float32
)

print("Stroke input:",
      train_inputs["stroke_input"].shape)

print("Labels:",
      train_inputs["labels"].shape)

print("Stroke input:",
      val_inputs["stroke_input"].shape)

print("Labels:",
      val_inputs["labels"].shape)
```

```python
history = ctc_model.fit(
    [
        X_train_padded,
        y_train_ctc,
        input_lengths_train_ctc,
        label_lengths_train_ctc
    ],
    train_dummy,

    validation_data=(
        [
            X_val_padded,
            y_val_ctc,
            input_lengths_val_ctc,
            label_lengths_val_ctc
        ],
        val_dummy
    ),

    batch_size=16,
    epochs=10,
    verbose=1
)
```
