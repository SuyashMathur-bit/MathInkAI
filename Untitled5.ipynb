```python
def extract_strokes(file_path):

    tree = ET.parse(file_path)

    root = tree.getroot()

    strokes = []

    for trace in root.iter():

        if trace.tag.endswith("trace"):

            if trace.text is None:
                continue

            points = []

            for point in trace.text.strip().split(","):

                values = point.strip().split()

                if len(values) >= 2:

                    x = float(values[0])
                    y = float(values[1])

                    points.append((x, y))

            if points:
                strokes.append(points)

    return strokes
```

```python
normalized_strokes = []

width = max_x - min_x
height = max_y - min_y

for stroke in strokes:

    normalized_stroke = []

    for x, y in stroke:

        new_x = (x - min_x) / width
        new_y = (y - min_y) / height

        normalized_stroke.append((new_x, new_y))

    normalized_strokes.append(normalized_stroke)

print("Number of strokes:", len(normalized_strokes))
print("First stroke:", normalized_strokes[0][:5])
```

```python
plt.figure(figsize=(6, 4))

for stroke in normalized_strokes:

    x = [point[0] for point in stroke]
    y = [point[1] for point in stroke]

    plt.plot(x, y)

plt.gca().invert_yaxis()

plt.title("Normalized Handwriting")
plt.xlabel("Normalized X")
plt.ylabel("Normalized Y")

plt.show()
```
